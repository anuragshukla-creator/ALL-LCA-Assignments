```python
# To Remove an item from a specified index
list = [10, 20, 30, 40, 50]

index = int(input("Enter index to remove: "))

if 0 <= index < len(list):
    list.pop(index)
    print("Updated list:", list)
else:
    print("Invalid index")
```

```python
# TO  Find the acronym of a phrase
phrase = input("Enter a phrase: ")

words = phrase.split()
acronym = ""

for word in words:
    acronym += word[0].upper()

print("Acronym:", acronym)
```

```python
#TO Print abbreviation of a month given its number
month = int(input("Enter month number (1-12): "))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

if 1 <= month <= 12:
    print("Month abbreviation:", months[month - 1])
else:
    print("Invalid month number")
```

```python
#TO Merge two lists into a dictionary
keys = ["Name", "Age", "City"]
values = ["Rahul", 20, "Pune"]

my_dict = dict(zip(keys, values))

print("Dictionary:", my_dict)
```

```python
#TO Remove duplicate elements from a list
my_list = [10, 20, 10, 30, 20, 40, 30]

new_list = []

for value in my_list:
    if value not in new_list:
        new_list.append(value)

print("Original list:", my_list)
print("List without duplicates:", new_list)
```

```python
 #to Convert two lists into a dictionary using zip()
keys = ["Name", "Age", "Marks"]
values = ["Amit", 18, 85]

my_dict = dict(zip(keys, values))

print("Dictionary:", my_dict)
```

```python
# to Find common elements in two lists
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common = []

for value in list1:
    if value in list2 and value not in common:
        common.append(value)

print("Common elements:", common)
```
