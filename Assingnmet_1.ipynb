```python
#program to create a dictionary, tuple and list of students

#creating a list of students
students_list = [ "John", "Ron", "Harry", "Samantha", "David",  "Michael", ]

#creating a tuple of students
Students_tuple = ("John", "Ron", "Harry", "Samantha", "David",  "Michael", )

#creating a dictionary of students
Students_dict = {
    101: "John",
    102: "Ron",
    103: "Harry",
    104: "Samantha",
    105: "David",
    106: "Michael"
}
print("List of Students: ",students_list) 
print("Tuple of Students: ", Students_tuple)
print("Dictionary of Students: ",Students_dict) 

#1. Add operation to the dictionary
Students_dict[107] = "Alicent"
Students_dict[108] = "Renera"

print("\nAfter adding new students to the dictionary: ", Students_dict)

#2. Update operation to the dictionary
Students_dict[101] = "John Rai"
print("After updating student name in the dictionary: ", Students_dict)

#3. Delete operation to the dictionary
del Students_dict[102]
print("After deleting a student from the dictionary: ", Students_dict)







```

```python
#create a list of 5 integers

numbers = [10, 20, 30, 40, 50]
print("list of integers", numbers)

#access and individual element through index

print("First:", numbers[0])
print("second:", numbers[1])
print("Third:", numbers[2])
print("Fourth:", numbers[2])
print("Fifth:", numbers[4])

```

```python
#appending an element in a list

bikes =["duke", "Royal enfield", "Triumph", "Hayabusha"]
print(f"list of bikes", bikes)

bikes.append("zx10r")
print("Appended list of bikes",bikes)
```

```python
#reversing element in a list 
bikes =["duke", "Royal enfield", "Triumph", "Hayabusha"]
print(f"list of bikes", bikes)

bikes.reverse()
print("Reverse list of bikes",bikes)
```

```python
#inserting an element in a list
bikes =["duke", "Royal enfield", "Triumph", "Hayabusha"]
print(f"list of bikes", bikes)

bikes.insert(0, "Apache")
print("inserted list of bikes:",bikes)

```

```python
# to print the number of occurence of specified element in a list
bikes =["duke", "Royal enfield", "Triumph", "Hayabusha", "Royal enfield", "Triumph", "Hayabusha"]
print(f"list of bikes", bikes)

ocurrence = str(input("put in a fruit:"))
print(bikes.count(occurence))
```

```python
#joining list
bikes =["duke", "Royal enfield", "Triumph", "Hayabusha"]
print(f"list of bikes", bikes)

bikes2 =["Z900r", "Hero", "Scrambler", "bajaj"]
print(f"list of bikes", bikes2)

bikes2 = bikes + bikes2

print("new joining list of bikes", bikes2 )
```
